```python
mov=df[['id','title','genre','overview','tags']]
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv=CountVectorizer(max_features=10000,stop_words='english')
```

```python
cv
```

```python
vec=cv.fit_transform(mov['tags'].values.astype('U')).toarray()
```

```python
vec
```

```python
vec.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
sim=cosine_similarity(vec)
```

```python
def reommendation_System(movies):
  idx=mov[mov['title']==movies].index[0]
  distance=sorted(list(enumerate(sim[idx])),reverse=True,key=lambda x:x[1])
  for i in distance[0:5]:
    print(mov.iloc[i[0]].title)
```

```python
reommendation_System("The Godfather")
```

```python
import gradio as gr

def recommendation_System(movie):

    idx = mov[mov['title'] == movie].index[0]

    distance = sorted(
        list(enumerate(sim[idx])),
        reverse=True,
        key=lambda x: x[1]
    )

    recommendations = []

    for i in distance[1:6]:
        recommendations.append(mov.iloc[i[0]].title)

    return recommendations


def recommend(movie):

    result = recommendation_System(movie)

    return "\n".join(result)


demo = gr.Interface(
    fn=recommend,

    inputs=gr.Dropdown(
        choices=mov['title'].tolist(),
        label="🎬 Select Movie"
    ),

    outputs=gr.Textbox(
        label="🍿 Top 5 Recommendations",
        lines=6
    ),

    title="Movie Recommendation System",
    description="Select a movie and get similar recommendations using Cosine Similarity",

    theme=gr.themes.Soft()
)

demo.launch(share=True)
```
